# **Stop Word Removal**

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
```

```python
txt1= "Hello World. this is the sample text..."
```

```python
# Case folding

txt1=txt1.lower()
```

```python
txt1
```

```python
# Filter out stop word

new_sentence = [word for word in word_tokenize(txt1) if word not in stopwords.words('english')]
```

```python
new_sentence
```

```python
# Check stop words list from the given sentence

stop_word_list = [word for word in word_tokenize(txt1) if word in stopwords.words('english')]
stop_word_list
```

# **Stemming & Lemmatization**


```python
# Stemming
#  Stemming uses rigid rules to chop off word endings, which often leaves you with non-words (like happi)

import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
```

```python
# Initialize the stemmer
stemmer = PorterStemmer()
```

```python
# Sample Words

sample_words = ['running', 'runner', 'run', 'happiness','programming','bigger','biggest']
```

```python
# Applying stemming technique

stemmer.stem(sample_words[1])
```

```python
# Converting all

for result in sample_words:
  print(stemmer.stem(result))
```

```python
#Lemmatization
#Lemmatization uses a dictionary mapping to return a valid root word,
#transforming structural changes like plural to singular (mice to mouse) or past to present tense.

import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
sample_words_2 = ['geese','mice']
```

```python
# initialize
lemm = WordNetLemmatizer()
```

```python
lemm.lemmatize(sample_words_2[0])
```

```python
for i in sample_words_2:
  print(lemm.lemmatize(i))
```

## **Bag of Words (BoW)** --> Vectorization

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
# Dataset Spam/ham
df = pd.read_csv("spam.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.Category.value_counts()
```

```python
df.Category.value_counts()/len(df)*100
```

```python
# Converting ham as 0 and Spam as 1, adding as new column

df['spam'] = df['Category'].apply(lambda x: 1 if x=='spam' else 0)
```

```python
df.head()
```

## **Train Test Split**

```python
X_train,X_test,y_train,y_test = train_test_split(df.Message,df.spam,test_size=0.25)
```

```python
len(X_test)
```

```python
X_train[:4]
```

# Creating Bag of Words using countverctorization

```python
v = CountVectorizer()
X_train_cv = v.fit_transform(X_train.values)
X_test_cv = v.transform(X_test)
```

```python
X_train_cv
```

```python
# Just seeing in the array format of the X_train_cv, since its in integer format now

X_train_cv.toarray()
```

```python
v.vocabulary_ # count of words

#v.vocabulary_ maps each word to a unique column index in the feature matrix. For example,
#wrong: 7345 means the word wrong is represented by column 7345, not that it occurs 7,345 times.
```

```python
X_train_np = X_train_cv.toarray()
X_train_np[0]
np.where(X_train_np[0]!=0)
```

# **ML Model**

```python
# Using DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X_train_cv, y_train)
```

```python
y_pred = model.predict(X_test_cv)
```

```python
# Printing Classification Report

print(classification_report(y_test,y_pred))
```

```python
# Need to verify accurecy score

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)*100
```
